```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk import pos_tag
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
nltk.download("punkt")
nltk.download("stopwords")
nltk.download("averaged_perceptron_tagger")
nltk.download("wordnet")
```

```python
#Sample Document
text = "Text Analytics is very important in Data Science and Machine Learning."
```

```python
#Tokenization
tokens = word_tokenize(text)
print("Tokens: ", tokens)
```

```python
#POS Tagging
pos = pos_tag(tokens)
print("\nPos Tags: ", pos)
```

```python
#Stopwords Removal
stop_words = set(stopwords.words("english"))

filtered = [word for word in tokens if word.lower() not in stop_words]
print("\nAfter Stopword Removal: ", filtered)
```

```python
#Stemming
ps = PorterStemmer()

stemmed = [ps.stem(word) for word in filtered]
print("\nStemmed Words: ", stemmed)
```

```python
#Lemmatization
lm = WordNetLemmatizer()

lemmatized = [lm.lemmatize(word) for word in filtered]
print("\nLemmatized Words:", lemmatized)
```

```python
#TF - IDF
documents = [
    "Data Science is useful",
    "Machine Learning is part of Data Science",
    "Text Analytics uses Machine Learning"
]

tfidf = TfidfVectorizer()

result = tfidf.fit_transform(documents)

print("\nTF-IDF Matrix: ")
print(result.toarray())
```
